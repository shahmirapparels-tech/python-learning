## Question 7: Compute the factorial of a number

### Solution 1 — Using `for` loop

```python
number = int(input("Enter a non-negative integer: "))

if number < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1

    for i in range(1, number + 1):
        factorial *= i

    print("Factorial =", factorial)
```

### Solution 2 — Using `while` loop

```python
number = int(input("Enter a non-negative integer: "))

if number < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1
    i = 1

    while i <= number:
        factorial *= i
        i += 1

    print("Factorial =", factorial)
```

Demo password used in this assignment: `python123`. The `for` solution allows up to 1000 attempts, while the `while` solution continues until the correct password is entered.

## Question 11: Calculate the sum of even and odd numbers from 1 to 50

### Solution 1 — Using `for` loop

```python
even_sum = 0
odd_sum = 0

for i in range(1, 51):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print("Sum of even numbers =", even_sum)
print("Sum of odd numbers =", odd_sum)
```

### Solution 2 — Using `while` loop

```python
even_sum = 0
odd_sum = 0
i = 1

while i <= 50:
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i
    i += 1

print("Sum of even numbers =", even_sum)
print("Sum of odd numbers =", odd_sum)
```

## Question 12: Create a guest list and prevent duplicate names

The assignment specifically requires the `for` version to use a list, `input()`, and `if-else`.

### Solution 1 — Using `for` loop

```python
guest_list = []

for i in range(10):
    name = input("Enter guest name: ").strip()

    if name in guest_list:
        print("Duplicate name! This guest is already on the list.")
    else:
        guest_list.append(name)
        print("Guest added successfully.")

print("\nFinal Guest List:")
for guest in guest_list:
    print(guest)
```

### Solution 2 — Using `while` loop

```python
guest_list = []
i = 0

while i < 10:
    name = input("Enter guest name: ").strip()

    if name in guest_list:
        print("Duplicate name! This guest is already on the list.")
    else:
        guest_list.append(name)
        print("Guest added successfully.")

    i += 1

print("\nFinal Guest List:")
for guest in guest_list:
    print(guest)
```

## Question 13: Create a restaurant menu while excluding "fish"

The assignment specifically asks for the `for` solution to use `if`, `continue`, and a list to exclude `"fish"`.

### Solution 1 — Using `for` loop

```python
menu = []

for i in range(5):
    dish = input("Enter dish name: ").strip().lower()

    if dish == "fish":
        print("Fish is not available. Skipping this dish.")
        continue

    menu.append(dish)

print("\nRestaurant Menu:")
for dish in menu:
    print(dish)
```

### Solution 2 — Using `while` loop

```python
menu = []
i = 0

while i < 5:
    dish = input("Enter dish name: ").strip().lower()

    if dish == "fish":
        print("Fish is not available. Skipping this dish.")
        i += 1
        continue

    menu.append(dish)
    i += 1

print("\nRestaurant Menu:")
for dish in menu:
    print(dish)
```

## Question 14: Dawat menu and cost calculation for 300 guests

This question requires:
- 300 guests
- `discard_menu = ["fish", "qeema", "nihari"]`
- 5 dish inputs
- rejected dishes skipped with `continue`
- `menu_price` dictionary
- per-head cost = total dish price / 300
- `menu_price_per_head` dictionary
- final total menu price and complete per-head cost.

### Solution 1 — Using `for` loop

```python
number_of_guests = 300
discard_menu = ["fish", "qeema", "nihari"]
menu = []
menu_price = {}
menu_price_per_head = {}

for i in range(5):
    dish = input("Enter dish name: ").strip().lower()

    if dish in discard_menu:
        print("This dish is not allowed. Skipping...")
        continue

    price = float(input(f"Enter total price for {number_of_guests} guests: "))

    menu.append(dish)
    menu_price[dish] = price

for dish in menu:
    menu_price_per_head[dish] = menu_price[dish] / number_of_guests

total_menu_price = sum(menu_price.values())
complete_per_head_cost = sum(menu_price_per_head.values())

print("\n----- Dawat Menu Summary -----")
print("Number of guests:", number_of_guests)
print("Allowed dishes:", menu)
print("Menu prices:", menu_price)
print("Per head costs:", menu_price_per_head)
print("Total menu price:", total_menu_price)
print("Complete per head cost:", complete_per_head_cost)
```

### Solution 2 — Using `while` loop

```python
number_of_guests = 300
discard_menu = ["fish", "qeema", "nihari"]
menu = []
menu_price = {}
menu_price_per_head = {}

i = 0

while i < 5:
    dish = input("Enter dish name: ").strip().lower()

    if dish in discard_menu:
        print("This dish is not allowed. Skipping...")
        i += 1
        continue

    price = float(input(f"Enter total price for {number_of_guests} guests: "))

    menu.append(dish)
    menu_price[dish] = price

    i += 1

for dish in menu:
    menu_price_per_head[dish] = menu_price[dish] / number_of_guests

total_menu_price = sum(menu_price.values())
complete_per_head_cost = sum(menu_price_per_head.values())

print("\n----- Dawat Menu Summary -----")
print("Number of guests:", number_of_guests)
print("Allowed dishes:", menu)
print("Menu prices:", menu_price)
print("Per head costs:", menu_price_per_head)
print("Total menu price:", total_menu_price)
print("Complete per head cost:", complete_per_head_cost)
```
