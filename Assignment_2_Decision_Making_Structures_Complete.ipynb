### Question 1: Check if a given number is positive, negative or zero.

```python
# Program 1
num = 15.5  # Input number
print(f"Input Number: {num}")

if num > 0:
    print("Result: The number is Positive.")
elif num < 0:
    print("Result: The number is Negative.")
else:
    print("Result: The number is Zero.")
```

### Question 7: Basic Calculator (/, x, +, -).

```python
# Program 7
n1 = 50.0
op = 'x'
n2 = 4.0
print(f"Calculation: {n1} {op} {n2}")

if op == '+':
    print("Result:", n1 + n2)
elif op == '-':
    print("Result:", n1 - n2)
elif op == 'x' or op == '*':
    print("Result:", n1 * n2)
elif op == '/':
    if n2 != 0:
        print("Result:", n1 / n2)
    else:
        print("Error: Division by zero")
else:
    print("Invalid operator")
```

### Question 12: Tax Deduction (5% for Salary > 600,000).

```python
# Program 12
salary1 = 800000.0
print(f"Input Salary: Rs. {salary1:,.2f}")

if salary1 > 600000:
    tax = salary1 * 0.05
    gross_salary = salary1 - tax
    print(f"Tax Deducted (5%): Rs. {tax:,.2f}")
    print(f"Gross Salary: Rs. {gross_salary:,.2f}")
else:
    print(f"Tax Deducted: Rs. 0.00")
    print(f"Gross Salary: Rs. {salary1:,.2f}")
```

### Question 13: Salary Slabs & Tax Deduction
- > 2,500,000 ---> 15%
- 1,500,000 - 2,500,000 ---> 10%
- 500,000 - 1,500,000 ---> 5%
- < 500,000 ---> 0%

```python
# Program 13
salary2 = 1800000.0
print(f"Input Salary: Rs. {salary2:,.2f}")

if salary2 > 2500000:
    tax_rate = 0.15
elif salary2 >= 1500000:
    tax_rate = 0.10
elif salary2 >= 500000:
    tax_rate = 0.05
else:
    tax_rate = 0.0

tax_amount = salary2 * tax_rate
gross_sal = salary2 - tax_amount

print(f"Applied Tax Rate: {tax_rate * 100}%")
print(f"Tax Deducted: Rs. {tax_amount:,.2f}")
print(f"Gross Salary: Rs. {gross_sal:,.2f}")
```
